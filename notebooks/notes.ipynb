# 这是用拟合的IV反推的price，不是真实的price。你怎么保证自己的结果是可信的

（1）我在比较不同的IV，RV，VRP定义，对VRP策略的影响。虽然使用的是拟合的IV和价格，会存在回测误差，但我更关注的是相对的pnl，而不是绝对的pnl。计算相对PnL的时候，这些误差在所有定义都存在，所以影响相对可控。

（2）我之前想过另一套解释，但那一套有漏洞。我想的是，model surface会随着real surface移动，所以price或者iv变化的方向和幅度应该是类似的。但问题是，我可能只能说两个曲面相关性高，但是变化的幅度大小是不好说的，可能会有比较大的误差，对vrp这种吃小spread的策略的回测，影响比较大。而且从时间上来看，signal的产生，更可能在model overestimate iv的时候产生，这就会有系统误差。
